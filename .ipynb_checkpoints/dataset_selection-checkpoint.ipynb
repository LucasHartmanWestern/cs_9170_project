```python
from ucimlrepo import fetch_ucirepo 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score
```

```python

from ucimlrepo import fetch_ucirepo 
import pandas as pd

# Load UCI Appliance Energy dataset
appliances_energy_prediction = fetch_ucirepo(id=374)
df = pd.concat([appliances_energy_prediction.data.features,
                appliances_energy_prediction.data.targets], axis=1)

# Extract hour from timestamp for biasing
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d%H:%M:%S')
df['hour'] = df['date'].dt.hour
print(df.columns.tolist())
```

```python

def apply_bias(df, strategy):
    if strategy == 'none':
        return df.copy()
    elif strategy == 'time_of_day':
        return df[df['hour'] < 12].copy()  # keep morning, remove evening
    elif strategy == 'room_temperature':
        return df[df['T_out'] > 10].copy()  # remove cold outdoor temp
    elif strategy == 'cold_weather':
        return df[df['T_out'] < 5].copy()   # only cold weather
    elif strategy == 'low_usage':
        return df[df['Appliances'] < 100].copy()  # only low usage
    else:
        raise ValueError("Unknown bias strategy")
```

```python

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset
import numpy as np
import matplotlib.pyplot as plt

class LSTMRegressor(nn.Module):
    def __init__(self, input_dim, hidden_dim=32):
        super().__init__()
        self.lstm = nn.LSTM(input_dim, hidden_dim, batch_first=True)
        self.fc = nn.Linear(hidden_dim, 1)
    def forward(self, x):
        _, (hn, _) = self.lstm(x)
        return self.fc(hn[-1]).squeeze()

def to_sequences(X, y, seq_len=5):
    Xs, ys = [], []
    for i in range(len(X) - seq_len):
        Xs.append(X.iloc[i:i+seq_len].values)
        ys.append(y.iloc[i+seq_len])
    return torch.tensor(np.array(Xs), dtype=torch.float32), torch.tensor(np.array(ys), dtype=torch.float32)
```

```python
from sklearn.metrics import r2_score
import seaborn as sns

features = ['T1', 'RH_1', 'T2', 'RH_2', 'T_out']
target = 'Appliances'
strategies = ['none', 'time_of_day', 'room_temperature', 'cold_weather', 'low_usage']
results = []

for strategy in strategies:
    biased_df = apply_bias(df.copy(), strategy)
    if len(biased_df) < 100:
        print(f"Skipping {strategy}: not enough samples")
        continue

    X = biased_df[features]
    y = biased_df[target]
    X_seq, y_seq = to_sequences(X, y)

    # Trim date info to align with sequence
    biased_df = biased_df.iloc[5:].reset_index(drop=True)

    # Train/test split
    train_size = int(0.8 * len(X_seq))
    X_train, X_test = X_seq[:train_size], X_seq[train_size:]
    y_train, y_test = y_seq[:train_size], y_seq[train_size:]
    test_meta = biased_df.iloc[train_size:]

    # Train LSTM
    model = LSTMRegressor(input_dim=X.shape[1])
    optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
    criterion = nn.MSELoss()
    train_loader = DataLoader(TensorDataset(X_train, y_train), batch_size=64, shuffle=True)

    for epoch in range(20):
        for xb, yb in train_loader:
            optimizer.zero_grad()
            loss = criterion(model(xb), yb)
            loss.backward()
            optimizer.step()

    # Predict
    model.eval()
    with torch.no_grad():
        preds = model(X_test).numpy()
        y_true = y_test.numpy()

    if strategy == 'time_of_day':
        split = test_meta['hour'] < 12
        label1, label2 = 'Morning', 'Evening'
    elif strategy == 'room_temperature':
        split = test_meta['T_out'] > 10
        label1, label2 = 'Comfortable Temp', 'Cold Temp'
    elif strategy == 'cold_weather':
        split = test_meta['T_out'] < 2.5
        label1, label2 = 'Cold', 'Not Cold'
    elif strategy == 'low_usage':
        split = test_meta['Appliances'] < 100
        label1, label2 = 'Low Usage', 'High Usage'
    else:
        r2_all = r2_score(y_true, preds)
        results.append({'Strategy': strategy, 'Group': 'All', 'R2': r2_all})
        continue

    # Compare group R² only if both groups have at least 2 samples
    if np.sum(split) >= 2 and np.sum(~split) >= 2:
        r2_a = r2_score(y_true[split], preds[split])
        r2_b = r2_score(y_true[~split], preds[~split])
        results.append({'Strategy': strategy, 'Group': label1, 'R2': r2_a})
        results.append({'Strategy': strategy, 'Group': label2, 'R2': r2_b})
    else:
        print(f"Skipping {strategy}: One of the groups is too small.")
```

```python

result_df = pd.DataFrame(results)
plt.figure(figsize=(10, 5))
sns.barplot(data=result_df, x='Strategy', y='R2', hue='Group')
plt.title('LSTM R² Score by Bias Strategy (Well vs Underrepresented)')
plt.ylim(0, 1)
plt.grid()
plt.tight_layout()
plt.show()
```
